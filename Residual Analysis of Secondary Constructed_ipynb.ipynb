```
# Residual Analysis Using the Secondary Constructed Data Set

import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error

# Secondary constructed data set
observed = np.array([
    120, 145, 160, 180, 200,
    215, 230, 250, 270, 290,
    310, 330, 350, 370, 390,
    410, 430, 450, 470, 490,
    510, 530, 550, 570, 590,
    610, 630, 650, 670, 690
])

predicted = np.array([
    118, 150, 158, 175, 205,
    220, 225, 248, 275, 295,
    305, 335, 345, 375, 392,
    405, 435, 448, 475, 485,
    515, 528, 548, 575, 585,
    615, 625, 655, 668, 695
])

# Error statistics
residual = observed - predicted

rmse = np.sqrt(mean_squared_error(observed, predicted))
mae = mean_absolute_error(observed, predicted)
r2 = r2_score(observed, predicted)
bias = np.mean(residual)

# Mean Absolute Percentage Error (MAPE)
# Zero observed values are excluded because MAPE is undefined when O_i = 0.
nonzero_mask = observed != 0
mape = 100 * np.mean(
    np.abs((predicted[nonzero_mask] - observed[nonzero_mask])
           / observed[nonzero_mask])
)

# Factor-of-Two Fraction (FAC2)
# FAC2 is the fraction of non-zero observed values for which
# 0.5 <= P_i / O_i <= 2.
ratio = predicted[nonzero_mask] / observed[nonzero_mask]
fac2 = np.mean((ratio >= 0.5) & (ratio <= 2.0))

print("----------------------------------------")
print("Residual Analysis Results")
print("----------------------------------------")
print(f"R²   = {r2:.4f}")
print(f"RMSE = {rmse:.4f}")
print(f"MAE  = {mae:.4f}")
print(f"MAPE = {mape:.2f}%")
print(f"FAC2 = {fac2:.4f} ({fac2 * 100:.2f}%)")
print(f"Bias = {bias:.4f}")
print("----------------------------------------")
# Figure 1: Secondary Constructed Dataset vs. Model-Predicted Concentrations
plt.figure(figsize=(6, 6))
plt.scatter(observed, predicted)

xmin = min(observed.min(), predicted.min())
xmax = max(observed.max(), predicted.max())

plt.plot([xmin, xmax], [xmin, xmax], '--')
plt.xlabel('Secondary Constructed Dataset Concentration')
plt.ylabel('Model-Predicted Concentration')
plt.title('Figure 1. Secondary Constructed Dataset vs. Model-Predicted Concentrations')
plt.grid(True)
plt.tight_layout()
plt.show()

# Figure 2: Residuals vs. Model-Predicted Concentrations
plt.figure(figsize=(6, 5))
plt.scatter(predicted, residual)
plt.axhline(0, linestyle='--')
plt.xlabel('Model-Predicted Concentration')
plt.ylabel('Residual')
plt.title('Figure 2. Residuals vs. Model-Predicted Concentrations')
plt.grid(True)
plt.tight_layout()
plt.show()

# Figure 3: Distribution of Residuals
plt.figure(figsize=(6, 5))
plt.hist(residual, bins=8)
plt.xlabel('Residual')
plt.ylabel('Frequency')
plt.title('Figure 3. Distribution of Residuals')
plt.grid(True)
plt.tight_layout()
plt.show()

# Figure 4: Residuals Across Sampling Points
plt.figure(figsize=(8, 4))
plt.plot(
    range(1, len(residual) + 1),
    residual,
    marker='o'
)
plt.axhline(0, linestyle='--')
plt.xlabel('Sampling Point')
plt.ylabel('Residual')
plt.title('Figure 4. Residuals Across Sampling Points')
plt.grid(True)
plt.tight_layout()
plt.show()
```
