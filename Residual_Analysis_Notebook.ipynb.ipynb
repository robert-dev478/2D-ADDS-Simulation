# Residual Analysis
This notebook computes error statistics and produces diagnostic plots.

```
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error
```

## Input data

```
observed=np.array([120,145,160,180,200,215,230,250,270,290,310,330,350,370,390,410,430,450,470,490,510,530,550,570,590,610,630,650,670,690])
predicted=np.array([118,150,158,175,205,220,225,248,275,295,305,335,345,375,392,405,435,448,475,485,515,528,548,575,585,615,625,655,668,695])
```

## Error statistics

```
residual=observed-predicted
rmse=np.sqrt(mean_squared_error(observed,predicted))
mae=mean_absolute_error(observed,predicted)
r2=r2_score(observed,predicted)
bias=np.mean(residual)
print(r2,rmse,mae,bias)
```

## Diagnostic plots

```
# Observed vs Predicted
plt.figure(figsize=(6,6));plt.scatter(observed,predicted)
xmin=min(observed.min(),predicted.min());xmax=max(observed.max(),predicted.max())
plt.plot([xmin,xmax],[xmin,xmax],'--');plt.grid();plt.show()

# Residual vs Predicted
plt.figure(figsize=(6,5));plt.scatter(predicted,residual);plt.axhline(0,ls='--');plt.grid();plt.show()

# Histogram
plt.figure(figsize=(6,5));plt.hist(residual,bins=8);plt.grid();plt.show()

# Residual by Sampling Point
plt.figure(figsize=(8,4));plt.plot(range(1,len(residual)+1),residual,marker='o');plt.axhline(0,ls='--');plt.grid();plt.show()
```
